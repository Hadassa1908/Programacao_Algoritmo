```python
#Aumento
salario = input ("Entre com seu salário atual: ")
salario = float(salario)
percentual = input ("Entre com o percentual de aumento: ")
percentual = float(percentual)
aumento = salario*(percentual/100)
novo = salario+aumento
print("O aumento será de: ", aumento)
print("Seu novo salário é de: ", novo)
```
